## MATPLOTLIB

## Histogram

### 1.Basic Histogram

```python
import matplotlib.pyplot as plt

blood_sugar=[113,85,90,150,149,88,82,
             129,93,115,135,80,77]

plt.hist(blood_sugar)

plt.title("Blood Sugar Histogram")
plt.xlabel("Blood Sugar Level")
plt.ylabel("Frequency")

plt.show()
```

### 2.Histogram using bins and width

```python
import matplotlib.pyplot as plt

blood_sugar=[113,85,90,150,149,88,82,
             129,93,115,135,80,77]

plt.hist(blood_sugar,
         rwidth=0.5,
         bins=5,
         color="yellow")

plt.title("Blood Sugar Chart")
plt.xlabel("Blood Sugar Level")
plt.ylabel("Number of patients")

plt.show()
```

### 3.Histogram with Normal,Pre-diabetic,Diabetic

```python
import matplotlib.pyplot as plt

blood_sugar=[113,85,90,150,149,88,82,
             129,93,115,135,80,77]

plt.title("Blood Sugar Chart")
plt.xlabel("Blood Sugar Level")
plt.ylabel("Number of patients")

plt.hist(blood_sugar,
         rwidth=0.5,
         bins=[80,100,125,150],
         color="r")

plt.xticks([90,112,137],
           ["Normal\n(80-100)",
            "Pre-diabetic\n(100-125)",
            "Diabetic\n(125-150)"])

plt.show()
```

### 4.Multiple Data Sample Histogram

```python
import matplotlib.pyplot as plt
blood_sugar_men=[113,85,90,150,149,88,82,
             129,93,115,135,80,77]

blood_sugar_women=[67,98,89,120,133,150,84,
                   69,89,79,120,112,100]

plt.title("Blood Sugar Comparison")
plt.xlabel("Blood Sugar Level")
plt.ylabel("Number of patients")

plt.hist([blood_sugar_men,blood_sugar_women],
         rwidth=0.3,
         bins=[60,80,100,125,150],
         color=["red","green"],
         label=["Men","Women"])

plt.legend()
plt.show()
```

### 5.Step type Histogram

```python
import matplotlib.pyplot as plt

blood_sugar=[113,85,90,150,149,88,82,
             129,93,115,135,80,77]

plt.hist(blood_sugar,
         bins=[80,100,125,150],
         histtype='step',
         color="b")

plt.title("Blood Sugar Chart")
plt.xlabel("Blood Sugar Level")
plt.ylabel("Number of patients")

plt.show()
```

### 6.Horizontal Histogram

```python
import matplotlib.pyplot as plt

blood_sugar=[113,85,90,150,149,88,82,
             129,93,115,135,80,77]

plt.hist(blood_sugar,
         rwidth=0.4,
         bins=[80,100,125,150],
         color="b",
         orientation='horizontal')

plt.title("Blood Sugar Chart")
plt.ylabel("Blood Sugar Level")
plt.xlabel("Number of patients")

plt.show()
```

## SCATTER PLOT

### 1.Basic scatter plot

```python
import matplotlib.pyplot as plt

x=[5,7,8,7,2,17,2,9,4,11]
y=[99,86,87,88,100,86,103,87,94,78]

plt.scatter(x,y,
            color="red",
            marker='x',
            s=50,
            alpha=0.7)

plt.xlabel("x values")
plt.ylabel("y values")
plt.title("Basic Scatter Plot")

plt.show()
```

### 2.Coloured Scatter plot Based on values

```python
import matplotlib.pyplot as plt
import numpy as np

x=np.random.rand(50)
y=np.random.rand(50)

colors=np.random.rand(50)

plt.scatter(x,y,
            c=colors,
            cmap='viridis',
            s=100)

plt.colorbar()
plt.title('Color mapped scatter plot')

plt.show()
```

### 3.Bubble chart Style Scatter plot

```python
import matplotlib.pyplot as plt
import numpy as np

x=[5,7,8,7,2,17,2,9,4,11]
y=[99,86,87,88,100,86,103,87,94,78]

sizes=1000 * np.random.rand(10)
colors=np.random.rand(10)

plt.scatter(x,y,
            s=sizes,
            c=colors,
            cmap='plasma',
            alpha=0.6,
            edgecolors='white')


plt.colorbar()
plt.title('Bubble chart style scatter plot')
plt.xlabel("x values")
plt.ylabel("y values")

plt.show()
```

### 4.Scatter plot using pandas

```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.DataFrame({"height":[160,170,180,175,165],
                 "weight":[55,65,75,70,60]})

plt.scatter(df["height"],
            df["weight"])

plt.xlabel("Height(cm)")
plt.ylabel("Weight(kg)")
plt.title("Height Vs Weight")

plt.show()
```

## PIE CHART

### 1.Basic Pie Chart

```python
import matplotlib.pyplot as plt

exp_vals=[1400,600,300,410,250]
exp_labels=["Home rent","Food","Bill","Car","Other utilities"]

plt.pie(exp_vals,
        labels=exp_labels)
plt.axis('equal')

plt.show()
```

### 2.Pie chart with percentage,shadow and exploded slices

```python
import matplotlib.pyplot as plt

exp_vals=[1400,600,300,410,250]
exp_labels=["Home rent","Food","Bill","Car","Other utilities"]

plt.pie(exp_vals,
        labels=exp_labels,
        shadow=True,
        autopct='%0.1f%%',
        radius=2,
        explode=[0.5,0.3,0.4,0.2,0.2])

plt.axis('equal')

plt.show()
```

### 3.Pie chart in clockwise direction

```python
import matplotlib.pyplot as plt

exp_vals=[1400,600,300,410,250]
exp_labels=["Home rent","Food","Bill","Car","Other utilities"]

plt.pie(exp_vals,
        labels=exp_labels,
        shadow=True,
        autopct='%0.1f%%',
        radius=2,
        counterclock=False,
        explode=[0,0,0,0.2,0.2],
        startangle=90)

plt.axis('equal')

plt.show()
```

### 4.Save pie chart in pdf format

```python
import matplotlib.pyplot as plt

exp_vals=[1400,600,300,410,250]
exp_labels=["Home rent","Food","Bill","Car","Other utilities"]

plt.pie(exp_vals,
        labels=exp_labels,
        shadow=True,
        autopct='%0.1f%%',
        radius=2,
        counterclock=False,
        explode=[0,0,0,0.2,0.2])

plt.axis('equal')

plt.savefig("expenses-pie.pdf",
            bbox_inches='tight',
            pad_inches=3)

plt.show()
```
