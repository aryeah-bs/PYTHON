## MATPLOTLIB-DATA VISUALISATION IN PYTHON

### Line plots

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[10,12,8,4,11]
plt.plot(x,y)

plt.title("Line plot example")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6]
y=[50,51,52,48,47,46]
plt.plot(x,y,color="indigo",linewidth=3,linestyle=":")

plt.title("weather")
plt.xlabel("Day")
plt.ylabel("Temperature")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6]
y=[50,51,52,48,47,46]
plt.plot(x,y,"g+:")

plt.title("weather")
plt.xlabel("Day")
plt.ylabel("Temperature")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6]
y=[50,51,52,48,47,46]
plt.plot(x,y,color="#00FF00",
                    marker="*",linestyle='-')

plt.title("weather")
plt.xlabel("Day")
plt.ylabel("Temperature")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6]
y=[50,51,52,48,47,46]
plt.plot(x,y,"r+-",markersize=10)
plt.show()
```

```python
#alpha property
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6]
y=[50,51,52,48,47,46]
plt.plot(x,y,"b<-",alpha=0.5)
plt.show()
```

```python
#legend
import matplotlib.pyplot as plt
days=[1,2,3,4,5,6,7]

max_t=[50,51,52,43,47,49,46]
min_t=[43,42,40,44,33,35,37]
avg_t=[45,48,48,46,40,42,41]

plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Weather")

plt.plot(days,max_t,label="Maximum")
plt.plot(days,min_t,label="Minimum")
plt.plot(days,avg_t,label="Average")

plt.legend(loc="best",shadow=True,fontsize="large")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
days=[1,2,3,4,5,6,7]

max_t=[50,51,52,43,47,49,46]
min_t=[43,42,40,44,33,35,37]
avg_t=[45,48,48,46,40,42,41]

plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Weather")

plt.plot(days,max_t,label="Maximum")
plt.plot(days,min_t,label="Minimum")
plt.plot(days,avg_t,label="Average")

plt.legend(loc="upper right",shadow=True,fontsize="large")
plt.grid(True)
plt.show()
```

### Bar charts

```python
#simple bar chart
import matplotlib.pyplot as plt
import numpy as np

company=["Google","Amazon","Flipkart","Facebook"]
revenue=[90,136,70,27]

a=np.arange(len(company))

plt.bar(a,revenue,label="Revenue")
plt.xticks(a,company)
plt.ylabel("Revenue")
plt.title("Tech Company Stocks")

plt.legend()
plt.show()
```

```python
#multiple bar charts
import matplotlib.pyplot as plt
import numpy as np

company=["Google","Amazon","Flipkart","Facebook"]
revenue=[90,136,70,27]
profit=[40,10,34,12]

a=np.arange(len(company))

plt.bar(a,revenue,label="Revenue")
plt.bar(a,profit,label="Profit")

plt.xticks(a,company)
plt.ylabel("Amount")
plt.title("Tech Company Stocks")

plt.legend()
plt.show()
```

```python
#seperate bar charts
import matplotlib.pyplot as plt
import numpy as np

company=["Google","Amazon","Flipkart","Facebook"]

revenue=[90,136,70,27]
profit=[40,10,34,12]

a=np.arange(len(company))

plt.bar(a - 0.2,revenue,width=0.4,label="Revenue",color="green")
plt.bar(a + 0.2,profit,width=0.4,label="Profit",color="red")

plt.xticks(a,company)
plt.ylabel("Amount")
plt.title("Tech Company Stocks")

plt.legend()
plt.show()
```

### Horizontal bar charts

```python
import matplotlib.pyplot as plt
import numpy as np

company=["Google","Amazon","Flipkart","Facebook"]

revenue=[90,136,70,27]
profit=[40,10,34,12]

a=np.arange(len(company))

plt.barh(a,revenue,label="Revenue")
plt.barh(a,profit,label="Profit")

plt.yticks(a,company)
plt.xlabel("Revenue")
plt.title("Tech Company Stocks")

plt.legend()
plt.show()
```

### Multiple barcharts using branch data

```python
import matplotlib.pyplot as plt
import numpy as np

Branch=["IT","ECE","CSE"]

IT=[12,30,18,22,15]
ECE=[28,6,16,5,10]
CSE=[29,3,24,25,17]

a=np.arange(len(IT))
barwidth=0.25

plt.bar(a - barwidth,IT,color="r",width=barwidth,
        edgecolor="grey",label="IT")

plt.bar(a,ECE,color="g",width=barwidth,
        edgecolor="grey",label="ECE")

plt.bar(a + barwidth,CSE,color="b",width=barwidth,
        edgecolor="grey",label="CSE")

plt.xlabel("Branch")
plt.ylabel("Students passed")
plt.title("Students Passed in Different Branches")

plt.legend()
plt.show()
```
