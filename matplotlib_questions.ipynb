```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd 

np.random.seed(10)

#create dataset
students=['A','B','C','D','F','G']
maths=np.random.randint(40,100,6)
science=np.random.randint(40,100,6)

#create dataframe
df=pd.DataFrame({"Students":students,
                 "Maths":maths,
                 "Science":science})

print('STUDENT DATASET:\n',df)

#Line chart
plt.figure()
plt.plot(df["Students"],df["Maths"],marker='o',label="Maths")
plt.plot(df["Students"],df["Science"],marker='s',label="Science")

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Students mark trend")

plt.legend()
plt.grid(True)

plt.show()

#bar charts
plt.figure()
a=np.arange(len(students))

plt.bar(a - 0.2,df['Maths'],width=0.4,label="Maths")
plt.bar(a + 0.2,df['Science'],width=0.4,label="Science")

plt.xticks(a,students)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Maths Vs Science marks")

plt.legend()

plt.show()

#scatter plot
plt.figure()
plt.scatter(df["Maths"],
            df["Science"],
            c=df['Science'],
            cmap='magma')

plt.colorbar(label="Science mark")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Maths Vs Science marks")

plt.show()

#pie chart
total_marks = df["Maths"]

plt.figure()
plt.pie(total_marks,
        labels=df["Students"],
        autopct='%1.1f%%')

plt.title("Maths Marks Distribution")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(0)

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = np.random.randint(2000, 10000, 6)
profit = np.random.randint(500, 3000, 6)

#Create DataFrame (Pandas)
df = pd.DataFrame({
    "Month": months,
    "Sales": sales,
    "Profit": profit})

print("DATASET:\n", df)

#LINE CHART (Sales Trend)
plt.figure()
plt.plot(df["Month"], df["Sales"], marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
    
#BAR CHART (Sales vs Profit)
plt.figure()
x = np.arange(len(months))

plt.bar(x - 0.2, df["Sales"], width=0.4, label="Sales")
plt.bar(x + 0.2, df["Profit"], width=0.4, label="Profit")

plt.xticks(x, months)
plt.title("Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()
plt.show()


# SCATTER PLOT (Relationship)
plt.figure()
plt.scatter(df["Sales"], df["Profit"], c=df["Profit"], cmap='magma')

plt.colorbar(label="Profit Level")
plt.title("Sales vs Profit Relationship")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()

#PIE CHART (Total Distribution)
total_sales = df["Sales"]

plt.figure()
plt.pie(total_sales, labels=df["Month"], autopct='%1.1f%%')
plt.title("Sales Distribution by Month")
plt.show()
```
