```python
#determinent
import numpy as np
A=np.array([[1,2],
           [3,4]])
print(np.linalg.det(A))
```

```python
#matrix inverse
import numpy as np
A=np.array([[1,2],
           [3,4]])
print(np.linalg.inv(A))
```

```python
#eigen values and eigen vectors
import numpy as np
A=np.array([[1,2],
           [3,4]])
values,vectors=np.linalg.eig(A)
print("Eigenvalues:",values)
print("Eigenvectors:\n",vectors)
```

```python
#solve linear equations
import numpy as np
A=np.array([[1,2],
            [2,4]])
b=np.array([8,13])
x=np.linalg.solve(A,b)
print("solution:",x)
```
