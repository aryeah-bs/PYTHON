```python
class LibraryBook:
    def __init__(self, book):
        self.book = book
        self.borrowed = False
    def borrow(self):
        if not self.borrowed:
            self.borrowed = True
            print(self.book, "has been borrowed.")
        else:
            print(self.book, "is already borrowed.")
     def return_book(self):
        if self.borrowed:
            self.borrowed = False
            print(self.book, "has been returned.")
        else:
            print(self.book, "was not borrowed.")


book = LibraryBook("Mathilukal")
book.borrow()
book.return_book()
```

### Types of inheritance
### Single inhritance

```python
class Parent:
    def func1(self):
        print("This function is in parent class")
class Child(Parent):
    def func2(self):
        print("This function is in child class")
obj=Child()
obj.func1()
obj.func2()
```

### Multiple inheritance

```python
class Mother:
    mothername=""
    def mother(self):
        print(self.mothername)
class Father:
    fathername=""
    def father(self): 
        print(self.fathername)
class Son(Mother,Father):
    def parents(self):
        print("Father:",self.fathername)
        print("Mother:",self.mothername)
s1=Son()
s1.fathername="RAM"
s1.mothername="SITA"
s1.parents()
```

### Multilevel inheritance

```python
class Grandfather:
    def __init__(self,grandfathername):
        self.grandfathername=grandfathername
        
class Father(Grandfather):
    def __init__(self,fathername,grandfathername):
        self.fathername=fathername
        Grandfather.__init__(self,grandfathername)
        
class Son(Father):
    def __init__(self,sonname,fathername,grandfathername):
        self.sonname=sonname
        Father.__init__(self,fathername,grandfathername)

    def print_name(self):
        print("Grandfather:",self.grandfathername)
        print("Father:",self.fathername)
        print("Son:",self.sonname)

s1=Son("Prince","Rampal","Lal mani")
print(s1.grandfathername)
s1.print_name()
```

### Hierarchical inheritance

```python
class Parent:
    def func1(self):
        print("This function is in parent class")
        
class Child1(Parent):
    def func2(self):
        print("This function is in child1")

class Child2(Child1):
    def func3(self):
        print("This function is in child2")

obj1=Child1()
obj2=Child2()

obj1.func1()
obj1.func2()

obj2.func1()
obj2.func2()
obj2.func3()
        
```

```python
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.__marks=marks #private attribute
        
    def get_marks(self):
        return self.__marks

    def set_marks(self,marks):
        if marks>=0 and marks<=100:
            self.__marks=marks
        else:
            print("Invalid marks")

s=Student("Arya",90)
print(s.name)
print(s.get_marks())
s.set_marks(95)
print(s.get_marks())

    
```

```python
###ATM
def atm():
    balance=1000
    pin="1234"
    print("...WELCOME TO THE ATM...")
    pin_no=input("Enter the 4 digit PIN number:")
    if pin_no!=pin:
        print("Incorrect PIN.Exiting...")
        return

    while True:
        print("\nATM Menu:")
        print("1.Check balance")
        print("2.Deposit")
        print("3.Withdraw")
        print("4.Exit")
        choice=input("Enter your choice 1-4:")
        if choice=='1':
            print(f"Your current balance is: ${balance}")
        elif choice=='2':
            amount=int(input("Enter the amount to deposit:"))
            if amount>0:
                balance+=amount
                print(f"Successfully deposited ${amount} New balance:${balance}")

            else:
                print("Invalid input")

        elif choice=='3':
            amount=int(input("Enter the amount to withdraw:"))
            if amount<=balance:
                balance-=amount
                print(f"Successfully withdrawn ${amount} New balance:${balance}")

            else:
                print("Insufficient funds")

        elif choice=='4':
            print("Thankyou for using the ATM....")
            break

        else:
            print("Invalid choice please select a valid option 1-4")

atm()

                
```

### Abstraction

```python
from abc import ABC,abstractmethod
class Bank(ABC):
    @abstractmethod
    def interest(self):
        pass

class SBI(Bank):
    def interest(self):
        print("Interest rate:6.5%")

class HDFC(Bank):
    def interest(self):
        print("Interest rate:7%")

s=SBI()
s.interest()

h=HDFC()
h.interest()
```

```python
from abc import ABC,abstractmethod
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("car starts with a key")

class Bike(Vehicle):
    def start(self):
        print("Bike starts with a self-start button")

c=Car()
c.start()

b=Bike()
b.start()
```
