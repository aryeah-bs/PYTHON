### Try except

```python
try:
    print(b)

except:
    print("An exception occured")
```

```python
try:
    print(b)
except NameError:
    print("variable b is not defined")
except:
    print("something else went wrong")
    
```

```python
try:
    print("hello")
except NameError:
    print("something went wrong")
else:
    print("nothing went wrong")
```

```python
try:
    print(b)
except NameError:
    print("something went wrong")
finally:
    print("The 'try except' is finished")
```
