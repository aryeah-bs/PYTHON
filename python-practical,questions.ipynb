### Student management system

```python
students = {}
def add(students):
    n = int(input("How many students do you want to add? "))

    for i in range(n):
        print("\nEnter details for Student", i + 1)

        name = input("Student Name: ")
        student_id = int(input("Student ID: "))

        maths = int(input("Maths Mark: "))
        english = int(input("English Mark: "))
        computer = int(input("Computer Mark: "))

        students[student_id] = {
            "name": name,
            "maths": maths,
            "english": english,
            "computer": computer,
        }

        print("Student added successfully!")

def display(students):
    if not students:
        print("Student details are not found")
        return

    for student_id, details in students.items():
        print("\n===== Student Details =====")
        print("Student Name:", details["name"])
        print("Student ID:", student_id)
        print("Maths:", details["maths"])
        print("English:", details["english"])
        print("Computer:", details["computer"])

def search(students):
    student_id = int(input("Enter Student ID: "))

    if student_id in students:
        details = students[student_id]

        print("\n===== Student Found =====")
        print("Student Name:", details["name"])
        print("Student ID:", student_id)
        print("Maths:", details["maths"])
        print("English:", details["english"])
        print("Computer:", details["computer"])

    else:
        print("Student not found")

def update(students):
    student_id = int(input("Enter Student ID: "))

    if student_id in students:
        name = input("Enter new name: ")

        maths = int(input("Enter new Maths mark: "))
        english = int(input("Enter new English mark: "))
        computer = int(input("Enter new Computer mark: "))

        students[student_id]["name"] = name
        students[student_id]["maths"] = maths
        students[student_id]["english"] = english
        students[student_id]["computer"] = computer

        print("Student updated successfully!")

    else:
        print("Student not found.")


def calculate(students):
    if not students:
        print("No student details found.")
        return

    print("\n===== Student Results =====")

    for student_id, details in students.items():

        total = (
            details["maths"]
            + details["english"]
            + details["computer"]
        )

        average = total / 5

        if average >= 90:
            grade = "A+"
        elif average >= 80:
            grade = "A"
        elif average >= 70:
            grade = "B"
        elif average >= 60:
            grade = "C"
        elif average >= 50:
            grade = "D"
        else:
            grade = "F"

        print("\nStudent ID:", student_id)
        print("Student Name:", details["name"])
        print("Total:", total)
        print("Average:", round(average, 2))
        print("Grade:", grade)
def delete(students):
    student_id=int(input("Enter  student ID to delete:"))

    if student_id in students:
        del students[student_id]
        print("Student detils delete successfully")
    else:
        print("Student not found")
        
while True:
    print("\n===== Student Management System =====")
    print("1. Add Student")
    print("2. Display Students")
    print("3. Search Student")
    print("4. Update Student")
    print("5. Calculate Results")
    print("6. Delete Student")
    print("7. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add(students)
    elif choice == "2":
        display(students)
    elif choice == "3":
        search(students)
    elif choice == "4":
        update(students)
    elif choice == "5":
        calculate(students)
    elif choice == "6":
        delete(students)
    elif choice == "7":
        print("Thank you!")
        break
    else:
        print("Invalid choice. Please try again.")   

```
