### Recursive function

```python
def fibbnocci(x):
    if x==0:
        return 0
    elif x==1:
        return 1
    else:
        return fibbnocci(x-1)+fibbnocci(x-2)
print(fibbnocci(10))
```

```python
def fact(n):
    if n==0:
        return 1
    else:
        return n*fact(n-1)
print(fact(5))
```

### Local and Global variables

```python
def my_fun():
    a=1
    print(a)
my_fun()
```

```python
a=50
def my_fun():
    print("a inside function:",a)
my_fun()
print("a outside function:",a)
```

```python
a=50
def my_fun():
    global a
    a=2
my_fun()
print("a inside function",a)
print("a outside function",a)
```

```python
a=50
def my_fun():
    global a
    a=a+1
    print("a inside function",a)
my_fun()
print("a outside function",a)
```
